```python
import pandas as pd
import numpy as np

df = pd.read_csv("Hospital_Management_Cleaned.csv")

print("Dataset loaded successfully")
print("Total rows:", df.shape[0])
print("Total columns:", df.shape[1])
```

```python
print("Column Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nFirst 5 Records:")
display(df.head())

print("\nStatistical Summary:")
display(df.describe(include="all").T)
```

```python
print("Missing Values:")
display(df.isnull().sum())

print("\nDuplicate Records:", df.duplicated().sum())

print("\nUnique Values:")
display(df.nunique())
```

```python
categorical_cols = df.select_dtypes(
    include=["object", "category"]
).columns

for col in categorical_cols:
    print(f"\nAnalysis of {col}")
    display(df[col].value_counts().head(10))
```

```python
numeric_cols = df.select_dtypes(
    include=np.number
).columns

display(
    df[numeric_cols].agg(
        ["count", "mean", "median", "min", "max", "std"]
    ).T
)
```

```python
for col in categorical_cols:
    print(f"\nTop categories in {col}:")
    print(df[col].value_counts().head(5))
```

```python
correlation = df[numeric_cols].corr()

display(correlation.round(2))
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Corrected the file name to reference the existing CSV
df = pd.read_csv("Hospital_Management_Cleaned.csv")
```

```python
numeric_cols = df.select_dtypes(include=np.number).columns

if len(numeric_cols) > 0:
    col = numeric_cols[0]

    plt.figure(figsize=(8, 5))
    sns.histplot(data=df, x=col, kde=True)
    plt.title(f"Distribution of {col}")
    plt.show()
```

```python
categorical_cols = df.select_dtypes(
    include=["object", "category"]
).columns

if len(categorical_cols) > 0:
    col = categorical_cols[0]

    counts = df[col].value_counts().head(10)

    plt.figure(figsize=(9, 5))
    counts.plot(kind="bar")
    plt.title(f"Top Categories in {col}")
    plt.xlabel(col)
    plt.ylabel("Record Count")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()
```

```python
if len(numeric_cols) > 0:
    col = numeric_cols[0]

    plt.figure(figsize=(8, 4))
    sns.boxplot(x=df[col])
    plt.title(f"Box Plot of {col}")
    plt.show()
```

```python
if len(numeric_cols) >= 2:
    x_col = numeric_cols[0]
    y_col = numeric_cols[1]

    plt.figure(figsize=(8, 5))
    sns.scatterplot(data=df, x=x_col, y=y_col)
    plt.title(f"{x_col} vs {y_col}")
    plt.show()
```

```python
if len(numeric_cols) >= 2:
    plt.figure(figsize=(10, 7))

    sns.heatmap(
        df[numeric_cols].corr(),
        annot=True,
        cmap="coolwarm",
        fmt=".2f"
    )

    plt.title("Correlation Heatmap")
    plt.tight_layout()
    plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Select numerical columns
numeric_cols = df.select_dtypes(include="number").columns

# Line graph
if len(numeric_cols) > 0:
    column = numeric_cols[0]

    plt.figure(figsize=(10, 5))
    plt.plot(df.index, df[column], marker="o", linestyle="-")

    plt.title(f"Line Graph - {column}")
    plt.xlabel("Record Number")
    plt.ylabel(column)
    plt.grid(True)
    plt.tight_layout()
    plt.show()
else:
    print("No numerical columns found.")
```
