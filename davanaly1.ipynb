```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully")
```

```python
df = pd.read_csv('hospital_patient_management_dataset.csv')

df.head()
```

```python
# Display first 5 rows
display(df.head())

# Display last 5 rows
display(df.tail())

# Number of rows and columns
print("Dataset Shape:", df.shape)

# Column names
print("Column Names:")
print(df.columns)

# Dataset information
df.info()

# Statistical summary
df.describe(include='all')
```

```python
# Check missing values
print(df.isnull().sum())

# Check duplicate records
print("Duplicate Rows:", df.duplicated().sum())

# Remove extra spaces from text columns
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].str.strip()

# Remove duplicate rows
df = df.drop_duplicates()

# Check missing values again
print(df.isnull().sum())
```

```python
print("Total Patients:", len(df))

print("\nGender Distribution:")
print(df['Gender'].value_counts())

print("\nAverage Patient Age:")
print(df['Age'].mean())

print("\nDepartment-wise Patients:")
print(df['Department'].value_counts())
```

```python
diagnosis_count = df['Diagnosis'].value_counts()

print(diagnosis_count)
```

```python
department_count = df['Department'].value_counts()

print(department_count)
```

```python
print("Total Hospital Revenue:",
      df['Total_Bill_INR'].sum())

print("Average Hospital Bill:",
      df['Total_Bill_INR'].mean())
```
