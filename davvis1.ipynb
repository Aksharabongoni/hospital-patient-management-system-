```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (9, 5)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset to define df_clean
df_clean = pd.read_excel("/content/hospital_patient_management_dataset1.xlsx")

plt.figure(figsize=(7, 7))

gender_counts = df_clean["Gender"].value_counts()

plt.pie(
    gender_counts,
    labels=gender_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Patient Gender Distribution")
plt.show()
```

```python
plt.figure(figsize=(7, 7))

gender_counts = df_clean["Gender"].value_counts()

plt.pie(
    gender_counts,
    labels=gender_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Patient Gender Distribution")
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.histplot(
    data=df_clean,
    x="Age",
    bins=15,
    kde=True
)

plt.title("Patient Age Distribution")
plt.xlabel("Patient Age")
plt.ylabel("Number of Patients")
plt.show()
```

```python
plt.figure(figsize=(10, 5))

diagnosis_counts = df_clean["Diagnosis"].value_counts()

sns.barplot(
    x=diagnosis_counts.index,
    y=diagnosis_counts.values
)

plt.title("Common Diseases Among Patients")
plt.xlabel("Diagnosis")
plt.ylabel("Number of Patients")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 7))

status_counts = df_clean["Patient_Status"].value_counts()

plt.pie(
    status_counts,
    labels=status_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Patient Status Distribution")
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.histplot(
    data=df_clean,
    x="Total_Bill_INR",
    bins=20,
    kde=True
)

plt.title("Hospital Billing Distribution")
plt.xlabel("Total Bill (INR)")
plt.ylabel("Number of Patients")
plt.show()
```

```python
department_bill = df_clean.groupby(
    "Department"
)["Total_Bill_INR"].mean().sort_values(ascending=False)

plt.figure(figsize=(10, 5))

sns.barplot(
    x=department_bill.index,
    y=department_bill.values
)

plt.title("Average Hospital Bill by Department")
plt.xlabel("Department")
plt.ylabel("Average Bill (INR)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df_clean, y="Length_of_Stay_Days")

plt.title("Hospital Stay Duration")
plt.ylabel("Length of Stay (Days)")
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# Gender
df_clean["Gender"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    ax=axes[0, 0]
)
axes[0, 0].set_title("Gender Distribution")
axes[0, 0].set_ylabel("")

# Department
df_clean["Department"].value_counts().plot(
    kind="bar",
    ax=axes[0, 1]
)
axes[0, 1].set_title("Department-wise Patients")
axes[0, 1].tick_params(axis="x", rotation=45)

# Age
sns.histplot(
    df_clean["Age"].dropna(),
    kde=True,
    ax=axes[1, 0]
)
axes[1, 0].set_title("Patient Age Distribution")

# Patient status
df_clean["Patient_Status"].value_counts().plot(
    kind="bar",
    ax=axes[1, 1]
)
axes[1, 1].set_title("Patient Status")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.countplot(data=df_clean, x="Department")

plt.title("Department-wise Patient Count")
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig(
    "Hospital_Department_Visualization.png",
    dpi=300,
    bbox_inches="tight"
)

from google.colab import files
files.download("Hospital_Department_Visualization.png")
```
