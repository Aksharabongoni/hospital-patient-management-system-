```python
import pandas as pd
import numpy as np

df = pd.read_csv("Hospital_Management_1000_Rows_1000_Columns-1.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

df.head()
```

```python
# First 5 rows
display(df.head())

# Number of rows and columns
print("Rows and columns:", df.shape)

# All column names
print("Column names:")
print(df.columns.tolist())

# Data types and non-null values
df.info()

# Missing values in each column
print(df.isnull().sum())

# Duplicate rows
print("Duplicate rows:", df.duplicated().sum())
```

```python
# Keep a copy of the original data
df_clean = df.copy()

# 1. Remove extra spaces from column names
df_clean.columns = df_clean.columns.astype(str).str.strip()

# 2. Remove completely empty rows
df_clean = df_clean.dropna(how="all")

# 3. Remove completely empty columns
df_clean = df_clean.dropna(axis=1, how="all")

# 4. Remove extra spaces from text values
for col in df_clean.select_dtypes(include=["object", "string"]).columns:
    df_clean[col] = df_clean[col].str.strip()

# 5. Remove exact duplicate rows
df_clean = df_clean.drop_duplicates()

print("Cleaning completed!")
print("Original shape:", df.shape)
print("Cleaned shape:", df_clean.shape)
print("Duplicate rows remaining:", df_clean.duplicated().sum())
```

```python
missing_report = pd.DataFrame({
    "Missing_Count": df_clean.isnull().sum(),
    "Missing_Percentage": (
        df_clean.isnull().mean() * 100
    ).round(2)
})

missing_report = missing_report[
    missing_report["Missing_Count"] > 0
].sort_values("Missing_Percentage", ascending=False)

display(missing_report.head(30))
```

```python
# Check for missing values
print("Total missing values:", df_clean.isnull().sum().sum())

# Check duplicates
print("Duplicate rows:", df_clean.duplicated().sum())

# Check data types
print(df_clean.dtypes.value_counts())

# Preview cleaned dataset
display(df_clean.head(10))

# Summary of numerical columns
display(df_clean.describe())
```

```python
# Save cleaned dataset
output_file = "Hospital_Management_Cleaned.csv"

df_clean.to_csv(output_file, index=False)

print("Cleaned file saved:", output_file)
```
